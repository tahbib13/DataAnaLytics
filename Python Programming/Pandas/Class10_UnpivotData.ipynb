```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# UNPIVOT DATA

Here wide To Long Format Conversion

```python
df = pd.read_csv("Book2.csv")
df.head()
```

# Convert into LONG

```python

# Adjust settings to display all columns without truncation
#pd.set_option('display.max_columns', None)  # Display all columns
#pd.set_option('display.max_rows', None)     # Display all rows (if needed)
#pd.set_option('display.width', None)        # Make sure it fits to screen width

pd_long = pd.melt(df, id_vars=['Year', 'Month'], value_vars= ['1', '2', '3', '4'])
pd_long
```

# Now export the long format data .

```python
pd_long.to_csv("long.csv")
```

```python
pd_long.head(10)
```

# Pivoting in Python

```python
pd_wide = pd_long.pivot( index= ['Year', 'variable'], columns= 'Month',
                        values= 'value')

pd_wide.head(10)
```

# Task is calculate the missing value in month 9 and replace those missing value by median

```python

# Check for missing values in the month 9 column
missing_values = pd_wide[9].isna().sum()
print(f"Missing values in month 9: {missing_values}")

# Calculate the median for month 9, ignoring NaN values
median_month_9 = pd_wide[9].median()

# Replace missing values in month 9 with the median
pd_wide[9].fillna(median_month_9, inplace=True)

# Verify that there are no more missing values in month 9
missing_values_after = pd_wide[9].isna().sum()
print(f"Missing values in month 9 after replacement: {missing_values_after}")
print("Median values : {median_month_9}")

# Show the updated DataFrame (first 10 rows)
print(pd_wide.head(10))
```
