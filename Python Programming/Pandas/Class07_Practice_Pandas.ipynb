#Now we will see in financial sheet where currency wrote like $$,USD,Dollars, but our target is to convert one type currency so LIKE USD.

```python
import numpy as np
import pandas as pd

# Adjust settings to display all columns without truncation
pd.set_option('display.max_columns', None)  # Display all columns
pd.set_option('display.max_rows', None)     # Display all rows if needed
pd.set_option('display.width', None)        # Adjust display width automatically



def std_curr(x):
    if x == '$$' or x == 'Dollars':
        return "USD"
    return x 

df_financials = pd.read_excel("movies_db.xlsx", sheet_name= "financials",
                              
                              converters= {'currency': std_curr})
df_financials
```

#Data Joining

#INNER JOIN

#let assume client want to saw the movie financial situation

```python
import pandas as pd

# Adjust settings to display all columns without truncation
pd.set_option('display.max_columns', None)  # Display all columns
pd.set_option('display.max_rows', None)     # Display all rows if needed
pd.set_option('display.width', None)        # Adjust display width automatically

# Load the Excel file
file_path = 'movies_db.xlsx'

# Load the two sheets into separate DataFrames
df_movies = pd.read_excel(file_path, sheet_name='movies')
df_financials = pd.read_excel(file_path, sheet_name='financials')

# Perform an inner join on a common column (e.g., director_id)
merged_df = pd.merge(df_movies, df_financials, on='movie_id', how='inner')

# Display the merged DataFrame
merged_df
```

#So export the file

```python
merged_df.to_excel("movies_merged.xlsx", sheet_name= "merged movies", index=False)
```

#LEFT JOIN

```python
merged_df02 = pd.merge(df_movies, df_financials, on='movie_id', how='left')
merged_df02
```

#Right Join

```python
merged_df03 = pd.merge(df_movies, df_financials, on='movie_id', how='right')
merged_df03
```
